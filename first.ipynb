```python
#hello world!
import csv
import math
from sklearn import tree
from sklearn import ensemble
from sklearn import metrics
from sklearn import datasets
from sklearn import svm
from sklearn.model_selection import ShuffleSplit
from sklearn.model_selection import train_test_split
from sklearn.model_selection import cross_val_score
from sklearn.model_selection import cross_val_predict
from sklearn.metrics import RocCurveDisplay
```

```python
print("hello world")
data = [] #list of dicts to hold the data
```

```python
with open('netflix_customer_churn.csv', newline='') as csvfile:
        reader = csv.DictReader(csvfile)
        for row in reader:
                data.append(row)
csvfile.close()
```

```python
print(data[0])
```

```python
#starting with a 70/10/20 split for train/tune/test
training_len = math.ceil(len(data) *.70)
tuning_end= math.ceil(len(data) *.80)
train_set = data[:training_len]
tuning_set = data[training_len:tuning_end]
test_set = data[tuning_end:]
```
